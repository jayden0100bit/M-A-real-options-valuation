# 04 · Comparable company analysis

Peers: **Comcast** (NBCUniversal inside a cable distributor), **AMC Networks** (pure-play cable networks) and **Netflix**
(streaming). Disney is shown as a reference only. Each is measured at the two valuation dates, using only SEC filings filed
on or before each date and Yahoo prices on the market-data date.

Limitations:
- **Small, imperfect peer set.** The natural peers (Time Warner, Viacom, CBS, Discovery, Scripps, Lionsgate) have no price
  history in the accessible source.
- **Share counts are a proxy.** They are diluted weighted-average shares for the latest quarter, not shares outstanding.
- **Netflix is excluded from the ranges.** Its EBITDA/EBIT are not comparable (content amortization sits in cost of
  revenues), and it is priced on growth.
- **May 2018 P/E is not meaningful.** LTM earnings include the TCJA deferred-tax remeasurements.
- **Forward multiples are unavailable.** No consensus estimates were collected.
