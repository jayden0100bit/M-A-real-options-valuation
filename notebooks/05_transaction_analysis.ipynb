# 05 · Transaction analysis: original offer, amended offer, closing

Deal terms come from the verified `transaction_assumptions.csv` (Stage 1). Value concepts stay separate:
headline equity value, headline enterprise ("transaction") value, market-implied equity value, cash and stock
consideration, and Disney's ASC 805 accounting purchase price.

The **unaffected price** is from 3 Nov 2017, the last trading day before press reports of the talks (May 2018 proxy
p.56). Premiums on the Disney consideration alone *understate* what 21CF holders received, because they also kept
New Fox. A closing-date package premium is shown separately and is nominal (not adjusted for market movement).
