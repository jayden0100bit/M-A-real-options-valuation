# 09 · Monte Carlo: probability that Disney created value

20,000 draws on the May 2018 information set. Each draw re-runs the Stage 3 DCF (Disney's forecasts for RemainCo, with
revenue-growth and margin shifts) and the Stage 4 synergy model (realization, delay, integration costs), then compares
the total with the price paid. All distributions are labelled estimates anchored to the Stage 3/4 ranges. With every
shock at zero, the engines reproduce the Stage 3 EV and the Stage 4 net synergy value (checked).

```python
from pathlib import Path
import subprocess, sys
import pandas as pd
from IPython.display import Image, display
pd.set_option("display.width", 220); pd.set_option("display.max_colwidth", 130); pd.set_option("display.max_rows", 300)
ROOT = Path.cwd().resolve()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
PROC, FIG = ROOT / "data" / "processed", ROOT / "docs" / "figures"
res = subprocess.run([sys.executable, str(ROOT / "src" / "valuation" / "build_stage5.py")], cwd=ROOT, capture_output=True, text=True, encoding="utf-8")
print(res.stdout.strip()); assert res.returncode == 0, res.stdout + res.stderr
V = pd.read_csv(PROC / "verification_checks.csv")

MI = pd.read_csv(PROC / "monte_carlo_inputs.csv"); MS = pd.read_csv(PROC / "monte_carlo_summary.csv")
MC = pd.read_csv(PROC / "monte_carlo_convergence.csv"); MT = pd.read_csv(PROC / "monte_carlo_tornado.csv"); SC = pd.read_csv(PROC / "monte_carlo_scenarios.csv")
```
